# 05a — Faster R-CNN (MobileNetV3 Large 320 FPN) Training

**Objective**:

Train **Faster R-CNN** (`fasterrcnn_mobilenet_v3_large_320_fpn`) on the ECUSTFD dataset using annotations from `Annotations_patched` (with fallback to original `Annotations`).

### COCO 1-Indexed Label Convention (Torchvision):
- Torchvision Faster R-CNN reserves **Index 0 for `__background__`**.
- The 20 target classes (19 foods + 1 reference coin) are assigned IDs from **`1` to `20`** (`apple=1, banana=2, ..., coin=5, ..., tomato=20`).

### Output & Artifact Structure:
- **COCO Annotations**: `data/processed/faster_rcnn_seg/new/annotations/{train.json, test.json}`
- **Deployed Checkpoint**: `models/faster_rcnn_new_best.pt` (loaded by notebook 06a/06b)
- **Detailed Run Artifacts**: `outputs/predictions/05_train_new_<ts>/best.pt`, `history.json`, `summary.txt`
- **Logs**: `outputs/logs/05a_faster_rcnn_train_<ts>.log` and `data/processed/faster_rcnn_seg/logs/train_run_new_<ts>.log`
